```
from unsloth import FastLanguageModel

# LÜTFEN DİKKAT: Drive'ınıza gidip "checkpoints" klasörüne bakın.
# İçindeki en yüksek numaralı klasörün adını (örneğin checkpoint-1600) aşağıdaki X yazan yere yazın.
checkpoint_yolu = "/content/drive/MyDrive/restaurant_data/checkpoints/checkpoint-2164"

print("Eğitilen model Drive'dan hafızaya geri yükleniyor...")
max_seq_length = 2048
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = checkpoint_yolu,
    max_seq_length = max_seq_length,
    dtype = None,
    load_in_4bit = True,
)

# Modeli hafızaya aldığımıza göre artık dışa aktarabiliriz
output_dir = "/content/drive/MyDrive/restaurant_data/qwen2_5_restaurant_critic"

print("Model birleştiriliyor ve GGUF formatına dönüştürülüyor...")
print("Bu işlem model boyutuna göre 10-15 dakika sürebilir, lütfen sekmeyi kapatmayın.")

model.save_pretrained_gguf(
    output_dir,
    tokenizer,
    quantization_method = "q4_k_m",
)

print(f"✅ Başarılı! Modeliniz Drive'daki '{output_dir}' klasörüne kaydedildi.")
```
